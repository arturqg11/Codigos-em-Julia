```julia
begin
    using Images
    using ImageFiltering
    using DSP
end
```

```julia
imagem_gato = load("Imagens/meowdy.jpg")
```

```julia
size(imagem_gato)
```

```julia
kernel = Kernel.gaussian((10,10))
```

```julia
function heatmap_kernel(kernel)
    to_rgb(x) = RGB(max(-x, 0), max(x, 0), 0)
    to_rgb.(kernel) / maximum(abs.(kernel))

end
```

```julia
heatmap_kernel(kernel)
```

```julia
imfilter(imagem_gato, kernel)
```

```julia
kernel_2 = [
    -0.5  -1.0  -0.5
    -1.0   7.0  -1.0
    -0.5  -1.0  -0.5
]
```

```julia
heatmap_kernel(kernel_2)
```

```julia
imfilter(imagem_gato, kernel_2)
```

```julia
kernel_3 = [
    -0.5  0  0.5
    -1    0  1
    -0.5  0  0.5
]
```

```julia
heatmap_kernel(kernel_3)
```

```julia
imfilter(imagem_gato, kernel_3)
```
